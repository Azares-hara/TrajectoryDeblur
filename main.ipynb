```python
import os
import numpy as np
from PIL import Image

data_dir = "/home/t24cs018/GOPRO_LARGE/"

global_min, global_max = 255, 0
count = 0

# Walk through all subdirectories and files
for root, dirs, files in os.walk(data_dir):
    for f in files:
        if f.lower().endswith((".png", ".jpg", ".jpeg")):
            path = os.path.join(root, f)
            img = Image.open(path)
            arr = np.array(img)

            # Update global min/max
            global_min = min(global_min, arr.min())
            global_max = max(global_max, arr.max())
            count += 1

print(f"Checked {count} images")
print("Raw pixel range across dataset:", global_min, global_max)

# Show what happens if normalized
print("After /255 normalization:", global_min/255.0, global_max/255.0)
print("After [-1,1] normalization:", (global_min-127.5)/127.5, (global_max-127.5)/127.5)
```

```python
import os
import cv2
import numpy as np
from glob import glob

root_dir = "/home/t24cs018/GOPRO_LARGE/train"

for seq in os.listdir(root_dir):
    sharp_dir = os.path.join(root_dir, seq, "sharp")
    if not os.path.isdir(sharp_dir):
        continue
    
    syn_blur_dir = os.path.join(root_dir, seq, "synthetic_blur")
    os.makedirs(syn_blur_dir, exist_ok=True)

    for sharp_path in glob(os.path.join(sharp_dir, "*.png")):
        img = cv2.imread(sharp_path)
        
        #Gaussian blur
        ksize = np.random.choice([3,5,7,9])
        blur_img = cv2.GaussianBlur(img, (ksize, ksize), 0)
        base = os.path.basename(sharp_path)
        cv2.imwrite(os.path.join(syn_blur_dir, base), blur_img)

print("Synthetic blur generated for all sequences.")
```
